## A store offers a 10% discount on a product that costs 1500. Write a Python expression to calculate the discounted price using the appropriate arithmetic operator. Ask the user to enter the product price and discount rate.

```python
product_price=int(input('Enter Product Price: '))
discount_rate=int(input('Enter Discount Price: '))
discount_expression=(product_price*discount_rate)/100
discounted_price=product_price-discount_expression
print(f'Discounted Price:{discounted_price}')
```
