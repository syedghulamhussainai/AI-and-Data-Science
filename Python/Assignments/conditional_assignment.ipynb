## A traffic system calculates fines:

Speed limit = 60 km/h
* If speed ≤ 60 → No fine
* If 61–80 → Fine = 500
* If 81–100 → Fine = 2000
* If >100 → License suspended
#### BUT:
*If emergency vehicle → no fine
*If rainy weather AND speed > 60 → fine increases by 50%

```python
speed = int(input("Enter Speed: "))
emergency = input("Is it an emergency vehicle? (yes/no): ")
weather = input("Enter Weather (rainy/normal): ")

if emergency == "yes":
    print("No Fine")

else:
    if speed <= 60:
        print("No Fine")

    elif speed <= 80:
        fine = 500

        if weather == "rainy" and speed > 60:
            fine = fine + (fine * 50 / 100)

        print(f"Fine: {fine}")

    elif speed <= 100:
        fine = 2000

        if weather == "rainy" and speed > 60:
            fine = fine + (fine * 50 / 100)

        print(f"Fine: {fine}")

    else:
        print("License Suspended")
```

## A company filters candidates:
* Experience ≥ 5 years AND skill score ≥ 80 → shortlisted
* Experience 2–4 years AND skill score ≥ 70 → interview
* Freshers → only if test score ≥ 85
* If criminal record = True → reject immediately
* If multiple job switches (>3 in 2 years) → flag as unstable

```python
experience = int(input("Enter Experience (years): "))
skill_score = int(input("Enter Skill Score: "))
test_score = int(input("Enter Test Score: "))
criminal_record = input("Criminal Record? (yes/no): ")
job_switches = int(input("Job Switches in last 2 years: "))

if criminal_record == "yes":
    print("Rejected")

else:
    if experience >= 5 and skill_score >= 80:
        print("Shortlisted")

    elif experience >= 2 and experience <= 4 and skill_score >= 70:
        print("Interview")

    elif experience == 0:
        if test_score >= 85:
            print("Freshers Eligible")
        else:
            print("Not Eligible")

    else:
        print("Not Eligible")

    if job_switches > 3:
        print("Flagged as Unstable")
```

## Write a program to decide passenger clearance status.
Take input:
* baggage_weight
* passport_valid (Yes/No)
* visa_status (Valid/Invalid)
* criminal_check (Clear/Flagged)
#### Rules:
* baggage > 30kg → extra charge
* passport invalid → reject
* visa invalid → reject
* criminal flagged → security hold
* otherwise → cleared for boarding

```python
baggage_weight = float(input("Baggage weight (kg): "))
passport_valid = input("Passport valid (Yes/No): ")
visa_status = input("Visa status (Valid/Invalid): ")
criminal_check = input("Criminal check (Clear/Flagged): ")

if passport_valid == "No":
    print("Reject")

elif visa_status == "Invalid":
    print("Reject")

elif criminal_check == "Flagged":
    print("Security Hold")

elif baggage_weight > 30:
    print("Extra Charge")

else:
    print("Cleared for Boarding")
```

## Write a program to detect suspicious banking transactions.
Take input:
* transaction_amount
* location_match (Yes/No)
* time_of_transaction(Day/Night)
* previous_transactions_count
* device_verified (Yes/No)
#### Rules:
* amount > 100,000 → suspicious
* location mismatch + high amount → fraud alert
* night transaction + unverified device → high risk
* many transactions (>10 in 1 hour) → freeze account
* otherwise → normal transaction

```python
transaction_amount = float(input("Transaction amount: "))
location_match = input("Location match (Yes/No): ")
time_of_transaction = input("Time of transaction (Day/Night): ")
previous_transactions_count = int(input("Previous transactions in 1 hour: "))
device_verified = input("Device verified (Yes/No): ")

if previous_transactions_count > 10:
    print("Freeze Account")

elif location_match == "No" and transaction_amount > 100000:
    print("Fraud Alert")

elif time_of_transaction == "Night" and device_verified == "No":
    print("High Risk")

elif transaction_amount > 100000:
    print("Suspicious Transaction")

else:
    print("Normal Transaction")
```

## Write a program to classify a person’s risk level in a police database.
Take input:
* criminal_record (Yes/No)
* number_of_cases
* travel_history_flagged (Yes/No)
* financial_activity_score
* last_activity_days
#### Rules:
* criminal_record = Yes AND cases > 3 → High Risk
* flagged travel + suspicious financial score → High Risk
* no activity for long time (>365 days) → Monitor
* clean record + normal activity → Low Risk
* mixed signals → Medium Risk

```python
criminal_record = input("Criminal record (Yes/No): ")
number_of_cases = int(input("Number of cases: "))
travel_history_flagged = input("Travel history flagged (Yes/No): ")
financial_activity_score = int(input("Financial activity score: "))
last_activity_days = int(input("Days since last activity: "))

if criminal_record == "Yes" and number_of_cases > 3:
    print("High Risk")

elif travel_history_flagged == "Yes" and financial_activity_score > 70:
    print("High Risk")

elif last_activity_days > 365:
    print("Monitor")

elif criminal_record == "No" and travel_history_flagged == "No" and financial_activity_score <= 70:
    print("Low Risk")

else:
    print("Medium Risk")
```
