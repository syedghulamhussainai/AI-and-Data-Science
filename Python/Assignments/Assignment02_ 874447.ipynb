# 7) Compound Interest: 
Write a Python program to calculate compound interest. Use the formula:  
Where: 
        A = P × (1 + r/n)ⁿ⁻ᵗ
        
        A= total amount 
        P = principal amount 
        r = annual interest rate (decimal) 
        n = number of times interest is compounded per year 
        t = time in years 
Take P, r, n, and t as inputs from the user.

```python
principal_amount=float(input('Enter Principal Amount here:'))
rate_decimal=float(input('Enter Annual Interest rate here:'))
num=int(input('Enter number of time compounded per year here:'))
time_year=float(input('Enter Time in Year here:'))
Total_amount=principal_amount*(1+rate_decimal/num)**(num*time_year)
print('Total Amount:',Total_amount)
```

the formula 
            
            F=G m1 m2/r2
    m1, m2 are the masses of the objects

# 19) Work Done by a Force:
Write a Python program to calculate the work done by a force. Use the formula: 
where: 


                W=F.D COS(theta)
                    W = work done 
                    f= force 
                    d = distance 
                    theta = angle between force and direction of movement (in degrees)
                    Take f, d, and θ as inputs from the user.

```python
import math
theta=int(input('Enter Theta Value here:'))
f=int(input('Enter Force here:'))
d=int(input('Enter Distance here:'))
w=f*d*math.cos(theta)
print('Work Done:',w)
```
